```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv(
    r"C:\Users\vishv\Downloads\archive (1)\Churn_Modelling.csv"
)

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)

display(df.head())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

df = df.drop(
    columns=["RowNumber", "CustomerId", "Surname"],
    errors="ignore"
)

print("\nUpdated Dataset Shape:", df.shape)
display(df.head())
```

```python
# Total customers who left and stayed
churn_counts = df["Exited"].value_counts()

print("Customer Churn Count:")
print(churn_counts)

# Churn percentage
churn_percentage = df["Exited"].value_counts(normalize=True) * 100

print("\nCustomer Churn Percentage:")
print(churn_percentage)

# Visualize churn distribution
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="Exited")

plt.title("Customer Churn Distribution")
plt.xlabel("Exited (0 = Stayed, 1 = Left)")
plt.ylabel("Number of Customers")

plt.show()
```

```python
# Churn by Gender
gender_churn = df.groupby("Gender")["Exited"].mean() * 100

print("Churn Rate by Gender:")
print(gender_churn)

# Churn by Geography
geo_churn = df.groupby("Geography")["Exited"].mean() * 100

print("\nChurn Rate by Geography:")
print(geo_churn)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.barplot(x=gender_churn.index, y=gender_churn.values, ax=axes[0])
axes[0].set_title("Churn Rate by Gender")
axes[0].set_ylabel("Churn Rate (%)")

sns.barplot(x=geo_churn.index, y=geo_churn.values, ax=axes[1])
axes[1].set_title("Churn Rate by Geography")
axes[1].set_ylabel("Churn Rate (%)")

plt.tight_layout()
plt.show()
```

```python
# Create Age Groups
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[18, 25, 35, 45, 55, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "55+"]
)

# Calculate churn rate
age_churn = df.groupby(
    "AgeGroup", observed=False
)["Exited"].mean() * 100

print("Churn Rate by Age Group:")
print(age_churn)

# Visualization
plt.figure(figsize=(8, 5))

sns.barplot(x=age_churn.index, y=age_churn.values)

plt.title("Customer Churn Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Churn Rate (%)")

plt.show()
```

```python
# Average Credit Score and Balance by Churn Status
customer_analysis = df.groupby("Exited")[
    ["CreditScore", "Balance"]
].mean()

print("Average Credit Score and Balance:")
display(customer_analysis)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.boxplot(data=df, x="Exited", y="CreditScore", ax=axes[0])
axes[0].set_title("Credit Score vs Churn")
axes[0].set_xlabel("Exited (0 = Stayed, 1 = Left)")

sns.boxplot(data=df, x="Exited", y="Balance", ax=axes[1])
axes[1].set_title("Account Balance vs Churn")
axes[1].set_xlabel("Exited (0 = Stayed, 1 = Left)")

plt.tight_layout()
plt.show()
```

```python
# Churn by Active Membership
active_churn = df.groupby("IsActiveMember")["Exited"].mean() * 100

print("Churn Rate by Active Membership:")
print(active_churn)

# Churn by Number of Products
products_churn = df.groupby("NumOfProducts")["Exited"].mean() * 100

print("\nChurn Rate by Number of Products:")
print(products_churn)

# Visualizations
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.barplot(x=active_churn.index, y=active_churn.values, ax=axes[0])
axes[0].set_title("Churn by Active Membership")
axes[0].set_xlabel("0 = Inactive, 1 = Active")
axes[0].set_ylabel("Churn Rate (%)")

sns.barplot(x=products_churn.index, y=products_churn.values, ax=axes[1])
axes[1].set_title("Churn by Number of Products")
axes[1].set_xlabel("Number of Products")
axes[1].set_ylabel("Churn Rate (%)")

plt.tight_layout()
plt.show()
```

```python
# Churn by Tenure
tenure_churn = df.groupby("Tenure")["Exited"].mean() * 100

print("Churn Rate by Tenure:")
print(tenure_churn)

# Churn by Credit Card
card_churn = df.groupby("HasCrCard")["Exited"].mean() * 100

print("\nChurn Rate by Credit Card:")
print(card_churn)

# Visualizations
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.barplot(x=tenure_churn.index, y=tenure_churn.values, ax=axes[0])
axes[0].set_title("Churn Rate by Tenure")
axes[0].set_xlabel("Years with Bank")
axes[0].set_ylabel("Churn Rate (%)")

sns.barplot(x=card_churn.index, y=card_churn.values, ax=axes[1])
axes[1].set_title("Churn Rate by Credit Card")
axes[1].set_xlabel("0 = No Card, 1 = Has Card")
axes[1].set_ylabel("Churn Rate (%)")

plt.tight_layout()
plt.show()
```

```python
# Select numerical columns
numeric_df = df.select_dtypes(include=["number"])

# Calculate correlation
correlation = numeric_df.corr()

print("Correlation with Customer Churn:")
print(correlation["Exited"].sort_values(ascending=False))

# Visualize correlation
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Customer Churn Correlation Heatmap")
plt.show()
```
